```python
import random

secret_number = random.randint(1,10)

number_of_attempts = 0

while number_of_attempts<=3:
  guessed_number = int(input("enter number to guess between 1-10: "))
  if guessed_number>secret_number:
    print("OOPS!.. its more than what i have");
  elif guessed_number<secret_number:
    print("OOPS!.. its less than what i have");
  else:
    print("Congratulations!.. this is what i have", secret_number);
    break;
  number_of_attempts += 1

if number_of_attempts==0:
   print("Sorry!.. you run out of expectations... this is what i have"+secret_number)



```
